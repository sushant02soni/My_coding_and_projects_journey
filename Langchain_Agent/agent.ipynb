```python
import os
import certifi
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_community.tools.tavily_search import TavilySearchResults

load_dotenv()
os.environ["SSL_CERT_FILE"] = certifi.where()


OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
search_tool = TavilySearchResults(max_results=5)
```

```python
from langchain.agents import create_agent
from langchain_classic import hub
```

```python
result=search_tool.invoke("Give me latest news on ai research papers")
result
```

```python
llm1= ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0,
    api_key=OPENAI_API_KEY
)
tools=[search_tool]
```

```python

prompt="You are a helpful assistant. Use tools when needed to answer questions accurately."
```

```python
agent=create_agent(
    model=llm1,
    tools=tools,
    system_prompt=prompt
)
#agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

response= agent.invoke({"input": "Give me latest news on ai research papers"})
print(response["output"]) 

```
